```python
import pandas as pd

ports = pd.read_csv("ports_master.csv")

print("Total ports:", len(ports))

print("Missing latitude:", ports["Latitude"].isna().sum())
print("Missing longitude:", ports["Longitude"].isna().sum())

print("\nLatitude range:")
print(ports["Latitude"].min(), "to", ports["Latitude"].max())

print("\nLongitude range:")
print(ports["Longitude"].min(), "to", ports["Longitude"].max())
```

```python
# Check for duplicate coordinates
duplicate_coordinates = (
    ports.groupby(["Latitude", "Longitude"])
    .size()
    .reset_index(name="count")
)

duplicate_coordinates = duplicate_coordinates[
    duplicate_coordinates["count"] > 1
]

print("Duplicate coordinate locations:", len(duplicate_coordinates))

if len(duplicate_coordinates) > 0:
    display(duplicate_coordinates.head(20))
```

```python
# Create unique weather locations from the global port database

weather_locations = (
    ports[
        [
            "World Port Index Number",
            "Main Port Name",
            "UN/LOCODE",
            "Country Code",
            "Latitude",
            "Longitude"
        ]
    ]
    .drop_duplicates(subset=["Latitude", "Longitude"])
    .reset_index(drop=True)
)

print("Total WPI ports:", len(ports))
print("Unique weather locations:", len(weather_locations))

display(weather_locations.head(10))
```

```python
weather_locations = (
    ports[
        [
            "World Port Index Number",
            "Main Port Name",
            "UN/LOCODE",
            "Country Code",
            "Latitude",
            "Longitude"
        ]
    ]
    .drop_duplicates(subset=["Latitude", "Longitude"])
    .reset_index(drop=True)
)

print("Total WPI ports:", len(ports))
print("Unique weather locations:", len(weather_locations))

display(weather_locations.head(10))
```

```python
# Save unique global weather locations

weather_locations.to_csv(
    "global_weather_locations.csv",
    index=False
)

print("Saved:", "global_weather_locations.csv")
print("Rows:", len(weather_locations))
print("Columns:", list(weather_locations.columns))
```

```python
# Map each port to its ERA5 atmospheric grid cell

weather_locations["era5_lat"] = (
    (weather_locations["Latitude"] * 4).round() / 4
)

weather_locations["era5_lon"] = (
    (weather_locations["Longitude"] * 4).round() / 4
)

unique_era5_cells = (
    weather_locations[
        ["era5_lat", "era5_lon"]
    ]
    .drop_duplicates()
    .reset_index(drop=True)
)

print("Unique weather locations:", len(weather_locations))
print("Unique ERA5 atmospheric grid cells:", len(unique_era5_cells))

display(unique_era5_cells.head(10))
```

```python
# Map each port to the ERA5 ocean-wave grid

weather_locations["wave_lat"] = (
    (weather_locations["Latitude"] * 2).round() / 2
)

weather_locations["wave_lon"] = (
    (weather_locations["Longitude"] * 2).round() / 2
)

unique_wave_cells = (
    weather_locations[
        ["wave_lat", "wave_lon"]
    ]
    .drop_duplicates()
    .reset_index(drop=True)
)

print("Unique weather locations:", len(weather_locations))
print("Unique ERA5 atmospheric cells:", len(unique_era5_cells))
print("Unique ERA5 wave cells:", len(unique_wave_cells))

display(unique_wave_cells.head(10))
```
