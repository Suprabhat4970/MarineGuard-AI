```python
required_cells = set(
    zip(
        largest_tile_cells["era5_lat"],
        largest_tile_cells["era5_lon"]
    )
)

downloaded_cells = set(
    (float(lat), float(lon))
    for lat in instant_ds.latitude.values
    for lon in instant_ds.longitude.values
)

matched_cells = required_cells.intersection(downloaded_cells)
missing_cells = required_cells - downloaded_cells

print("Required cells:", len(required_cells))
print("Downloaded cells:", len(downloaded_cells))
print("Matched cells:", len(matched_cells))
print("Missing cells:", len(missing_cells))
```

```python
import pandas as pd
import numpy as np

required_cells_df = largest_tile_cells[
    ["era5_lat", "era5_lon"]
].drop_duplicates().reset_index(drop=True)

rows = []

for _, cell in required_cells_df.iterrows():

    lat = cell["era5_lat"]
    lon = cell["era5_lon"]

    u10 = instant_ds["u10"].sel(
        latitude=lat,
        longitude=lon
    ).values

    v10 = instant_ds["v10"].sel(
        latitude=lat,
        longitude=lon
    ).values

    t2m = instant_ds["t2m"].sel(
        latitude=lat,
        longitude=lon
    ).values

    tp = accum_ds["tp"].sel(
        latitude=lat,
        longitude=lon
    ).values

    wind_speed = np.sqrt(u10**2 + v10**2)
    temperature_c = t2m - 273.15
    precipitation_mm = tp * 1000

    for i, time in enumerate(instant_ds["valid_time"].values):

        rows.append({
            "valid_time": time,
            "latitude": lat,
            "longitude": lon,
            "wind_speed_ms": wind_speed[i],
            "temperature_c": temperature_c[i],
            "precipitation_mm": precipitation_mm[i]
        })

tile_weather = pd.DataFrame(rows)

print("Rows:", len(tile_weather))
print("Unique cells:", tile_weather[["latitude", "longitude"]].drop_duplicates().shape[0])

display(tile_weather.head(10))
```
